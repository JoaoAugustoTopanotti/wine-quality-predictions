```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/group_3_winequality.csv", dtype={"alcohol": str})

df.describe().T
```

```python
maiores = df.sort_values("volatile acidity", ascending=False).head(20)
maiores = maiores[["volatile acidity", "type"]]
maiores["corrigido"] = maiores["volatile acidity"] / 1000
print(maiores)
```

```python
suspeitos = df[df["volatile acidity"].between(2, 50)]
print(suspeitos[["volatile acidity", "type"]])
```

```python
gemeas = df[
    (df["fixed acidity"] == 7.6)
    & (df["residual sugar"] == 4.2)
    & (df["total sulfur dioxide"] == 112)
]
print(gemeas)
```

```python
suspeitos = df[df["density"].between(2, 50)]
print(suspeitos[["density", "type"]])
```

```python
fora = df[(df["density"] > 1.05) & (df["density"] < 2)]
print(fora["density"].value_counts().sort_index())
```
